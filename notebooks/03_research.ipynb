## Experiment: Spectrogram Augmentation (Time/Frequency Masking)

**Motivation:** Simple spectrogram masking (SpecAugment-style) can improve generalisation by making the model robust to missing time–frequency regions.

**Method:** During training only, random time masks (up to 12 time bins, 2 masks per sample) and frequency masks (up to 2 frequency bins, 1 mask per sample) were applied to the input spectrogram slices. Validation and test data were not augmented.

**Result:** Overall test accuracy remained the same at **0.769** compared to the baseline model. However, the confusion matrix changed: performance on **Rupe A** improved slightly (11/15 correct vs 10/15 baseline), while **Rupe B** decreased slightly (9/11 correct vs 10/11 baseline). This suggests the augmentation introduced robustness that helped one class but may have reduced class-specific cues for the other, possibly due to the limited dataset size and very small frequency dimension (6 bins).

**Conclusion:** In this configuration, masking augmentation did not improve overall accuracy, but it altered class-wise behaviour. Future work could tune mask widths, apply augmentation probabilistically, or explore alternative features/architectures better suited to the small input dimensions.
